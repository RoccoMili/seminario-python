## [CONSIGNA]: Estadísticas de un texto

**Dado el siguiente texto almacenado en una variable, calcule e imprima:**
- La cantidad total de líneas.
- La cantidad total de palabras.
- El promedio de palabras por línea.
- Todas las líneas cuya cantidad de palabras esté por encima del promedio.

```python
text = """Beautiful is better than ugly.
Explicit is better than implicit.
Simple is better than complex.
Complex is better than complicated.
Flat is better than nested.
Sparse is better than dense.
Readability counts.
Special cases aren't special enough to break the rules.
Although practicality beats purity.
Errors should never pass silently.
Unless explicitly silenced.
In the face of ambiguity, refuse the temptation to guess.
There should be one-- and preferably only one --obvious way
to do it.
Although that way may not be obvious at first unless you're
Dutch.
Now is better than never.
Although never is often better than *right* now.
If the implementation is hard to explain, it's a bad idea.
If the implementation is easy to explain, it may be a good
idea.
Namespaces are one honking great idea -- let's do more of
those!"""
```

## [RESOLUCIÓN]:

```python
import sys

sys.path.append("../src")

import text_dissector as td

phrases = text.strip().split(".")

print(f"Total de líneas: {td.line_count(phrases)}")

print(f"Total de palabras: {sum(td.word_count(phrase) for phrase in phrases)}")

average = td.words_average(phrases)

print(f"Promedio de palabras por línea: {average:.2f}")

print(f"\nLíneas por encima del promedio ({average:.2f} palabras):")

over_average = [line for line in phrases if td.word_count(line) > average]

for phrase in over_average:
    print(f"- {phrase.strip()} ({td.word_count(phrase)} palabras)")
```
