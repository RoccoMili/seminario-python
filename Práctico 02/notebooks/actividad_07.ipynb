## [CONSIGNA]: Sorteo de amigo invisible
Solicite al usuario una lista de nombres de participantes (separados por
coma). El programa debe asignar aleatoriamente a cada persona un amigo
invisible, cumpliendo que **nadie se tenga a sí mismo**.

Validaciones:
- Debe haber al menos 3 participantes.
- No debe haber nombres duplicados (comparar ignorando mayúsculas y
espacios extra).

## [RESOLUCIÓN]:

```python
import sys

sys.path.append("../src")

import secret_santa as ss

friends = input("Ingrese los participantes (separados por coma):").split(",")

result = ss.randomizer(friends)
if result is None:
    print("Deben haber al menos 3 participantes y no pueden haber duplicados.")
else:
    print("Sorteo de amigo invisible:")
    for k, v in result.items():
        print(f"    {k} → {v}")
```
